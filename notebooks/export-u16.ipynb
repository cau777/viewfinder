# Export LiDAR tiles as compact u16 grids

Re-exports each original 1 km × 1 km Vancouver LiDAR tile as a raw stream of uint16 values, plus one shared `index.csv`.

**Per tile file `<NAME>.u16`**: `width * height` little-endian uint16 values, no header.
- Row-major and north-up: value `k` is at `row = k // width` and `col = k % width`. Row 0 is the northern edge and col 0 is the western edge.
- Each value is the maximum altitude in the cell (noise points and outliers above `max_altitude` excluded), scaled over the **global** range of the whole dataset:
  `alt = alt_min + v / 65534 * (alt_max - alt_min)` (metres). The step is `(alt_max - alt_min) / 65534`, which is 3.8 mm for -10.8 to 239.9 m, so the error is at most 1.9 mm.
- `65535` means the cell has no points.

**Per tile file `<NAME>.cls`**: the most common ASPRS class of each cell's points (same points as the altitude), 4 bits per cell, `width * height / 2` bytes, no header.
- Same cell order as `.u16`. Cell `k` is in byte `k // 2`: the low nibble for even `k`, the high nibble for odd `k`.
- Values 0-14 are the ASPRS class (1 unassigned, 2 ground, 3 low vegetation, 5 high vegetation, 6 building, 9 water, ...). Ties go to the lower class. Classes above 14 (towers, wire connectors, bridge decks; none occur in this dataset) are stored as 1.
- `15` means the cell has no points.
- Ocean cells are class 9 (water).
- Cells whose centre lies in the ocean (Marine Regions internal waters, `~/data/raw/world-internal-waters`) are set to 0 m whatever their points, so ships and empty water never block the view. Bridges over the water are flattened too.

**`index.csv`**: one row per tile.
- `x_start, y_start` / `x_end, y_end`: UTM (`crs`, EPSG:26910) centres of the first cell (NW, k = 0) and the last cell (SE, k = width*height-1).
  Cell centre: `x = x_start + col * (x_end - x_start) / (width - 1)`, `y = y_start + row * (y_end - y_start) / (height - 1)`.
- `lon_*, lat_*`: WGS84 coordinates of the four corner cell centres (`nw` = start, `se` = end). Bilinear interpolation between them gives lat/lon for any cell, accurate to about 2e-7° (≈ 2.5 cm, far below the 0.5 m cell size), so no projection library is needed.
- `alt_min, alt_max`: the global range used for scaling. It is the same on every row.
- `classes_file` is the tile's `.cls` file.
- `n_points` and `n_missing` are the points used and the empty cells (ocean cells are never missing).

Grids are aligned on the tile's nominal bounds (taken from its name, the SW corner), so neighbouring tiles join seamlessly even when a coastal tile has sparse data.

The export runs in two passes because the global range is only known once every tile has been gridded:
1. Grid each tile and keep it as a temporary float32 file.
2. Encode every temporary file with the global range, then delete it.

```python
import zipfile
import pathlib
import laspy
import numpy as np
import pandas as pd
import geopandas as gpd
import shapely
from pyproj import Transformer

base_path = pathlib.Path("~/").expanduser()
data_path = base_path / "data"
lidar_path = data_path / "raw" / "lidar-vancouver"
ocean_path = data_path / "raw" / "world-internal-waters" / "eez_internal_waters_v4.gpkg"
temp_path = base_path / "temp"
export_path = data_path / "lidar-vancouver-u16-classes"

index_df = pd.read_csv(lidar_path / "index.csv", sep=";")

tile_size = 1000  # metres; tile NAME is the SW corner "<x>_<y>" in EPSG:26910
resolution = 2000  # cells per side -> 0.5 m cells, 8 MB per tile
noise_classes = [7, 18]  # ASPRS low / high noise: would otherwise dominate the per-cell max
max_altitude = 240  # metres; points above are outliers (birds, antennas) that inflate the global range
ocean_altitude = 0  # metres; every cell centred in the ocean, so ships do not block the view
u16_missing = 65535
water_class = 9  # ASPRS; every ocean cell
unassigned_class = 1  # ASPRS; stands in for classes that do not fit in 4 bits
class_missing = 15  # 4-bit value of a cell without points

# Ocean polygon (Marine Regions internal waters) in the tiles' CRS, clipped to the LiDAR extent
tile_xy = index_df["NAME"].str.split("_", expand=True).astype(int)
lidar_extent = shapely.box(tile_xy[0].min(), tile_xy[1].min(), tile_xy[0].max() + tile_size, tile_xy[1].max() + tile_size)
ocean_bbox = gpd.GeoSeries([lidar_extent], crs="EPSG:26910").to_crs("EPSG:4326").total_bounds
ocean = gpd.read_file(ocean_path, bbox=tuple(ocean_bbox)).to_crs("EPSG:26910").clip(lidar_extent).union_all()
```

```python
def load_lidar(name):
    las_file = temp_path / f"{name}.las"
    try:
        # The uncompressed file does not need to live long
        with zipfile.ZipFile(lidar_path / f"{name}.zip", 'r') as zip_ref:
            zip_ref.extract(las_file.name, temp_path)
        las = laspy.read(las_file)
        keep = ~np.isin(np.asarray(las.classification), noise_classes) & (np.asarray(las.z) <= max_altitude)
        cls = np.asarray(las.classification)[keep]
        return np.asarray(las.x)[keep], np.asarray(las.y)[keep], np.asarray(las.z)[keep], cls, las.header.parse_crs()
    finally:
        las_file.unlink(missing_ok=True)


def grid_tile(name):
    """Max altitude and most common class per cell on a north-up (row 0 = north) resolution x resolution grid.
    Altitude NaN / class class_missing = no points; ocean cells are ocean_altitude and water_class."""
    x, y, z, cls, crs = load_lidar(name)
    x0, y0 = (int(v) for v in name.split("_"))
    cell = tile_size / resolution
    col = np.clip(np.floor((x - x0) / cell).astype(np.int64), 0, resolution - 1)
    row = np.clip(np.floor((y0 + tile_size - y) / cell).astype(np.int64), 0, resolution - 1)
    grid = np.full(resolution * resolution, -np.inf)
    np.maximum.at(grid, row * resolution + col, z)
    grid[np.isneginf(grid)] = np.nan
    grid = grid.reshape(resolution, resolution)
    classes = majority_class(row * resolution + col, cls)
    ocean = ocean_mask(x0, y0)
    grid[ocean] = ocean_altitude
    classes[ocean.ravel()] = water_class
    return grid, classes.reshape(resolution, resolution), crs, len(z)


def majority_class(cell, cls):
    """Most common class per cell (ties to the lower class), class_missing for cells without points.
    One bincount per class present keeps memory at a few arrays of resolution^2."""
    cells = resolution * resolution
    best_count = np.zeros(cells, np.int64)
    best = np.full(cells, class_missing, np.uint8)
    for c in np.unique(cls):  # ascending, and only a strictly higher count wins: ties keep the lower class
        count = np.bincount(cell[cls == c], minlength=cells)
        better = count > best_count
        best[better] = c if c < class_missing else unassigned_class
        best_count[better] = count[better]
    return best


def ocean_mask(x0, y0):
    """True for the cells whose centre lies in the ocean, on the same grid as grid_tile."""
    tile_ocean = shapely.clip_by_rect(ocean, x0, y0, x0 + tile_size, y0 + tile_size)
    if tile_ocean.is_empty:
        return np.zeros((resolution, resolution), dtype=bool)
    shapely.prepare(tile_ocean)
    centres = (np.arange(resolution) + 0.5) * tile_size / resolution
    cx, cy = np.meshgrid(x0 + centres, y0 + tile_size - centres)
    return shapely.contains_xy(tile_ocean, cx, cy)


def tile_metadata(name, grid, crs, n_points):
    x0, y0 = (int(v) for v in name.split("_"))
    half = tile_size / resolution / 2
    # Centres of the first (NW) and last (SE) cells
    x_start, y_start = x0 + half, y0 + tile_size - half
    x_end, y_end = x0 + tile_size - half, y0 + half
    corners = {"nw": (x_start, y_start), "ne": (x_end, y_start), "sw": (x_start, y_end), "se": (x_end, y_end)}
    transformer = Transformer.from_crs(crs, "EPSG:4326", always_xy=True)
    meta = {
        "name": name, "file": f"{name}.u16", "classes_file": f"{name}.cls", "crs": f"EPSG:{crs.to_epsg()}",
        "width": resolution, "height": resolution,
        "x_start": x_start, "y_start": y_start, "x_end": x_end, "y_end": y_end,
    }
    for key, (cx, cy) in corners.items():
        meta[f"lon_{key}"], meta[f"lat_{key}"] = transformer.transform(cx, cy)
    # Tile range, replaced by the global range once every tile is gridded
    has_data = not np.isnan(grid).all()
    meta["alt_min"] = float(np.nanmin(grid)) if has_data else np.nan
    meta["alt_max"] = float(np.nanmax(grid)) if has_data else np.nan
    meta["n_points"] = n_points
    meta["n_missing"] = int(np.isnan(grid).sum())
    return meta


def encode(grid, alt_min, alt_max):
    v = np.round((grid - alt_min) / (alt_max - alt_min) * (u16_missing - 1))
    return np.where(np.isnan(grid), u16_missing, np.clip(v, 0, u16_missing - 1)).astype("<u2")


def pack_classes(classes):
    """Two 4-bit classes per byte: even cells in the low nibble, odd cells in the high one."""
    flat = classes.ravel()
    return (flat[0::2] | (flat[1::2] << 4)).astype(np.uint8)


def unpack_classes(packed):
    return np.stack([packed & 0xF, packed >> 4], axis=1).ravel()


def decode(values, alt_min, alt_max):
    return np.where(values == u16_missing, np.nan, alt_min + values / (u16_missing - 1) * (alt_max - alt_min))
```

```python
# Try one tile and check the round-trip error (with the tile's own range here)
name = index_df["NAME"][0]
grid, classes, crs, n_points = grid_tile(name)
meta = tile_metadata(name, grid, crs, n_points)
decoded = decode(encode(grid, meta["alt_min"], meta["alt_max"]), meta["alt_min"], meta["alt_max"])
err = np.nanmax(np.abs(decoded - grid))
assert (unpack_classes(pack_classes(classes)) == classes.ravel()).all()
print("classes:", dict(zip(*np.unique(classes, return_counts=True))))
print(f"max abs error {err * 1000:.2f} mm (bound {(meta['alt_max'] - meta['alt_min']) / 65534 / 2 * 1000:.2f} mm)")
pd.Series(meta)
```

```python
import multiprocessing
from concurrent.futures import ProcessPoolExecutor, as_completed
from tqdm.auto import tqdm

export_all = True
# Each worker peaks at ~5 GB of RAM on the densest tiles; 12 workers got OOM-killed on the 52 GB server, and 4 leave room for the API
max_workers = 4
grids_path = temp_path / "lidar-grids"

def grid_to_temp(name):
    """Pass 1: grid a tile into a temporary float32 file, write its final class file, and return its metadata."""
    grid, classes, crs, n_points = grid_tile(name)
    meta = tile_metadata(name, grid, crs, n_points)
    # float32 keeps ~0.03 mm at these altitudes; encode() clips the rare value rounded past the range
    tmp_file = grids_path / f"{name}.f32.tmp"
    grid.astype("<f4").tofile(tmp_file)
    tmp_file.replace(grids_path / f"{name}.f32")
    # Classes need no global range: written in their final form straight away
    cls_file = export_path / meta["classes_file"]
    tmp_file = cls_file.with_suffix(".cls.tmp")
    pack_classes(classes).tofile(tmp_file)
    tmp_file.replace(cls_file)
    return meta

def run_parallel(function, names, description):
    results, failed = [], {}
    # fork so workers can see the functions defined in this notebook
    with ProcessPoolExecutor(max_workers, mp_context=multiprocessing.get_context("fork")) as pool:
        futures = {pool.submit(function, name): name for name in names}
        for future in tqdm(as_completed(futures), total=len(futures), desc=description):
            try:
                results.append(future.result())
            except Exception as e:
                failed[futures[future]] = repr(e)
    for name, err in failed.items():
        print(f"  {name}: {err}")
    return results, failed

if export_all:
    grids_path.mkdir(parents=True, exist_ok=True)
    export_path.mkdir(parents=True, exist_ok=True)
    temp_path.mkdir(parents=True, exist_ok=True)
    rows, failed = run_parallel(grid_to_temp, index_df["NAME"], "Gridding tiles")
    assert not failed, "fix the failed tiles before encoding: the global range needs all of them"
    out_index = pd.DataFrame(rows).sort_values("name").reset_index(drop=True)
    global_min, global_max = out_index["alt_min"].min(), out_index["alt_max"].max()
    print(f"global range {global_min} .. {global_max} m, max error {(global_max - global_min) / 65534 / 2 * 1000:.2f} mm")
```

```python
def encode_from_temp(name):
    """Pass 2: encode a temporary grid with the global range, then delete it."""
    temp_file = grids_path / f"{name}.f32"
    grid = np.fromfile(temp_file, dtype="<f4").astype(np.float64)
    out_file = export_path / f"{name}.u16"
    # Write to a temp file first so an interrupted run never leaves a partial tile behind
    tmp_file = out_file.with_suffix(".u16.tmp")
    encode(grid, global_min, global_max).tofile(tmp_file)
    tmp_file.replace(out_file)
    temp_file.unlink()
    return name

if export_all:
    export_path.mkdir(parents=True, exist_ok=True)
    _, failed = run_parallel(encode_from_temp, out_index["name"], "Encoding tiles")
    out_index["alt_min"], out_index["alt_max"] = global_min, global_max
    out_index.to_csv(export_path / "index.csv", index=False, float_format="%.12g")
    print(f"Done: {len(out_index) - len(failed)} ok, {len(failed)} failed")
```

```python
# Read back exactly as an external program would: only index.csv + the raw stream
index = pd.read_csv(export_path / "index.csv")
t = index.iloc[0]
v = np.fromfile(export_path / t["file"], dtype="<u2")
alt = np.where(v == 65535, np.nan, t["alt_min"] + v / 65534 * (t["alt_max"] - t["alt_min"]))
packed = np.fromfile(export_path / t["classes_file"], dtype=np.uint8)
cls = np.stack([packed & 0xF, packed >> 4], axis=1).ravel()
assert ((v == 65535) == (cls == 15)).all(), "a cell has an altitude but no class, or the reverse"
k = np.arange(v.size)
row, col = k // t["width"], k % t["width"]
x = t["x_start"] + col * (t["x_end"] - t["x_start"]) / (t["width"] - 1)
y = t["y_start"] + row * (t["y_end"] - t["y_start"]) / (t["height"] - 1)
# Bilinear lat/lon from the corner cells
u, w = col / (t["width"] - 1), row / (t["height"] - 1)
bilerp = lambda c: (1 - w) * ((1 - u) * t[f"{c}_nw"] + u * t[f"{c}_ne"]) + w * ((1 - u) * t[f"{c}_sw"] + u * t[f"{c}_se"])
lon, lat = bilerp("lon"), bilerp("lat")

# Compare against an exact projection
exact_lon, exact_lat = Transformer.from_crs(t["crs"], "EPSG:4326", always_xy=True).transform(x, y)
print("max lat/lon error (deg):", np.abs(lat - exact_lat).max(), np.abs(lon - exact_lon).max())
print("total size:", sum(f.stat().st_size for f in export_path.iterdir()) / 2**20, "MiB")
pd.DataFrame({"x": x, "y": y, "lat": lat, "lon": lon, "alt": alt, "class": cls}).dropna()
```
