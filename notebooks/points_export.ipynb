```python
# Try loading all datasets
import os
from pathlib import Path
import tqdm
import pandas as pd

BASE_PATH = Path('~/').expanduser()

files = os.listdir(BASE_PATH / "data/lidar-vancouver-altitude")
datasets = [(BASE_PATH / "data/lidar-vancouver-altitude" / file) for file in tqdm.tqdm(files)]
```

```python
min_lat = 999999999999999
min_lon = 999999999999999

max_lat = -99999999999999999
max_lon = -99999999999999999

for df_path in tqdm.tqdm(datasets):
    df = pd.read_parquet(df_path)
    min_lat = min(min_lat, df['lat'].min())
    min_lon = min(min_lon, df['lon'].min())
    max_lat = max(max_lat, df['lat'].max())
    max_lon = max(max_lon, df['lon'].max())


(min_lat, min_lon)
```

```python
lat_range = max_lat - min_lat
lon_range = max_lon - min_lon

(lat_range, lon_range)
```
